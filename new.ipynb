```python
import requests

url= "https://api-web.nhle.com/v1/standings/now"

response = requests.get(url)

data = response.json()

teams_data = []

for team in data['standings']:
      teams_data.append(dict(
          div_name = team['divisionName'],
          conf_name = team['conferenceName'],
          team_name = team['teamName']['default'],
          team_abbrv = team['teamAbbrev']['default'],
          logo = team['teamLogo']
      ))

teams_data
```

# Insert in to sql

```python
import pymysql

conn = pymysql.connect(
    host= "localhost",
    user= "root",
    password= "1234",
    database= "new_nhl"
)

cursor = conn.cursor()
```

```python
teams_data[0]
```

```python
cursor.execute("""CREATE TABLE teams (
               team_id INT AUTO_INCREMENT PRIMARY KEY,
               team_abbrev VARCHAR(10) UNIQUE,
               team_name VARCHAR(100),
               conference_name VARCHAR(100),
               division_name VARCHAR(100),
               logo_url TEXT)
               """)
```

```python
teams_data[0]
```

```python
insert_query = """INSERT INTO teams (team_abbrev,team_name,conference_name,division_name,logo_url)
                    VALUES(%s,%s,%s,%s,%s)"""

for i in teams_data:
   values = (i['team_abbrv'],
         i['team_name'],
         i['conf_name'],
         i['div_name'],
         i['logo'])
   cursor.execute(insert_query,values)

conn.commit()
```
